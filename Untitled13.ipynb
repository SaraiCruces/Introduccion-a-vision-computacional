```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import tensorflow_datasets as tfds
```

```python
os.makedirs("models", exist_ok=True)
```

```python
NOMBRES = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
           "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]
```

```python
def plot_learning_curve(history, titulo=""):
    plt.figure()
    plt.plot(history.history['accuracy'], label='Training accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation accuracy')
    plt.title(f'Accuracy Curves {titulo}')
    plt.legend()
    plt.show()

    plt.figure()
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title(f'Loss Curves {titulo}')
    plt.legend()
    plt.show()
```

```python
!pip install -q importlib_resources

import tensorflow_datasets as tfds
import tensorflow as tf

train_ds, val_ds, test_ds = tfds.load(
    'fashion_mnist',
    split=['train', 'test[:50%]', 'test[50%:]'],
    shuffle_files=True,
    as_supervised=True,
    batch_size=32,
)
def normalizar(x, y):
    return tf.cast(x, tf.float32) / 255.0, y

train_ds = train_ds.map(normalizar).prefetch(tf.data.AUTOTUNE)
val_ds   = val_ds.map(normalizar).prefetch(tf.data.AUTOTUNE)
test_ds  = test_ds.map(normalizar).prefetch(tf.data.AUTOTUNE)
```

```python
for imgs, labels in train_ds.take(1):
    plt.figure(figsize=(4, 4))
    plt.imshow(imgs[0].numpy().squeeze(), cmap="gray", vmin=0, vmax=1)
    plt.title(NOMBRES[labels[0].numpy()])
    plt.axis("off")
    plt.show()
```

```python
def construir_modelo(capas, dropout=0.0, lr=1e-3, con_2d=False):
    layers = [
        tf.keras.layers.Input(shape=(28, 28, 1)),
        tf.keras.layers.Flatten(),
    ]
    for u in capas:
        layers.append(tf.keras.layers.Dense(u, activation="relu"))
        if dropout > 0:
            layers.append(tf.keras.layers.Dropout(dropout))
    if con_2d:
        layers.append(tf.keras.layers.Dense(2, activation="linear", name="rep_2d"))
    layers.append(tf.keras.layers.Dense(10, activation="softmax"))

    model = tf.keras.Sequential(layers)
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=lr),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )
    return model
```

```python
configs = {
    "Config 1: 128-64 (base de clase)":     dict(capas=[128, 64],  dropout=0.0, lr=1e-3, con_2d=False),
    "Config 2: 256-128 + Dropout 0.3":      dict(capas=[256, 128], dropout=0.3, lr=1e-3, con_2d=False),
    "Config 3: 128 + capa 2D (rep_2d)":     dict(capas=[128],      dropout=0.0, lr=1e-3, con_2d=True),
}

resultados = []
modelos = {}
historias = {}

for i, (nombre, cfg) in enumerate(configs.items(), start=1):
    print(f"\n===== {nombre} =====")
    model = construir_modelo(**cfg)
    model.summary()

    ruta = f"models/best_config{i}.keras"
    checkpoint = tf.keras.callbacks.ModelCheckpoint(
        filepath=ruta, monitor="val_loss", mode="min",
        save_best_only=True, verbose=0,
    )

    history = model.fit(
        train_ds,
        epochs=15,
        validation_data=val_ds,
        callbacks=[checkpoint],
        verbose=1,
    )
    plot_learning_curve(history, titulo=f"- {nombre}")

    best = tf.keras.models.load_model(ruta)
    test_loss, test_acc = best.evaluate(test_ds, verbose=0)

    modelos[nombre] = best
    historias[nombre] = history
    resultados.append({
        "Configuración": nombre,
        "Val accuracy (mejor)": max(history.history["val_accuracy"]),
        "Test accuracy": test_acc,
        "Test loss": test_loss,
    })
```

```python
nombre_2d = "Config 3: 128 + capa 2D (rep_2d)"
modelo_2d = modelos[nombre_2d]

extractor = tf.keras.Model(
    inputs=modelo_2d.inputs,
    outputs=modelo_2d.get_layer("rep_2d").output,
)

val_vis = val_ds.take(94)  # ~3000 imágenes
puntos2d = extractor.predict(val_vis, verbose=0)
y_visual = np.concatenate([y.numpy() for _, y in val_vis])

plt.figure(figsize=(7, 6))
sc = plt.scatter(puntos2d[:, 0], puntos2d[:, 1], c=y_visual,
                 cmap="tab10", s=10, alpha=0.75)
cbar = plt.colorbar(sc, ticks=range(10))
cbar.ax.set_yticklabels(NOMBRES)
plt.title("Espacio latente 2D aprendido por el MLP (Fashion-MNIST)")
plt.xlabel("Neurona de representación 1")
plt.ylabel("Neurona de representación 2")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
df_resultados = pd.DataFrame(resultados)
display(df_resultados.round(4))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

mejor = df_resultados.sort_values("Test accuracy", ascending=False).iloc[0]["Configuración"]
modelo_mejor = modelos[mejor]

y_true = np.concatenate([y.numpy() for _, y in test_ds])
y_pred = np.argmax(modelo_mejor.predict(test_ds, verbose=0), axis=1)

fig, ax = plt.subplots(figsize=(10, 8))
ConfusionMatrixDisplay.from_predictions(
    y_true, y_pred, display_labels=NOMBRES, cmap="Blues", ax=ax, xticks_rotation=45
)
plt.title(f"Matriz de confusión - {mejor}")
plt.grid(False)
plt.show()
```
